```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc file dữ liệu sạch (hãy đảm bảo bạn đã upload file train_clean_phu.csv lên Colab)
from google.colab import files
uploaded = files.upload()

import pandas as pd
df = pd.read_csv('train_clean_phu.csv')
print(df.head())

# Kiểm tra nhanh các cột dữ liệu
print(df.head())

# Nếu dữ liệu chưa có cột 'Month', bạn có thể chuyển đổi từ cột ngày tháng (ví dụ: 'Order Date')
# df['Order Date'] = pd.to_datetime(df['Order Date'])
# df['Month'] = df['Order Date'].dt.month

# Gom nhóm tổng doanh thu theo Phân khúc (Segment) và Tháng (Month)
segment_seasonality = df.groupby(['Segment', 'Month'])['Sales'].sum().reset_index()

print("Bảng tổng hợp doanh thu theo tháng của từng phân khúc:")
print(segment_seasonality)

# Vẽ biểu đồ đường (Line chart) so sánh tính mùa vụ
plt.figure(figsize=(12, 6))
sns.lineplot(data=segment_seasonality, x='Month', y='Sales', hue='Segment', marker='o', linewidth=2.5)

plt.title('So sánh tính mùa vụ theo tháng giữa các phân khúc khách hàng', fontsize=14, fontweight='bold')
plt.xlabel('Tháng', fontsize=12)
plt.ylabel('Tổng doanh thu (Sales)', fontsize=12)
plt.xticks(range(1, 13))
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Phân khúc khách hàng')
plt.show()
```
